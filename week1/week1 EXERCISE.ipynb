To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from openai import OpenAI
from groq import Groq
from dotenv import load_dotenv
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
MODEL_GROQ = 'llama-3.3-70b-versatile'
```

```python
# set up environment
load_dotenv()

# OpenAI client
openai_client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# Ollama client (using OpenAI-compatible endpoint)
ollama_client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"  # Ollama doesn't require an API key
)

# Groq client
groq_client = Groq(api_key=os.getenv("GROQ_API_KEY"))
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get Groq to answer, with streaming
print("⚡ Groq (Llama 3.3 70B) says:\n")

stream = groq_client.chat.completions.create(
    model=MODEL_GROQ,
    messages=[
        {"role": "system", "content": "You are a helpful technical assistant who explains programming concepts clearly and concisely."},
        {"role": "user", "content": question}
    ],
    stream=True
)

for chunk in stream:
    if chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end='', flush=True)

print("\n")
```

```python
# Get Llama 3.2 to answer
print("🦙 Llama 3.2 says:\n")

response = ollama_client.chat.completions.create(
    model=MODEL_LLAMA,
    messages=[
        {"role": "system", "content": "You are a helpful technical assistant who explains programming concepts clearly and concisely."},
        {"role": "user", "content": question}
    ]
)

print(response.choices[0].message.content)
```
